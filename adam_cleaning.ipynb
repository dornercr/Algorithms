```python
import pandas as pd

# Load the CSV (change the filename accordingly)
df = pd.read_csv("transcripts_youtube_videos_levant_module11_with_ilr.csv", encoding='utf-8')

# ✅ Overwrite all dialects first
df['dialect'] = 'Levant'

# ✅ Normalize guessed_dialect and dialect for clean comparison
df['guessed_dialect_clean'] = df['guessed_dialect'].str.strip().str.lower()

# Handle common normalization issues (e.g., 'levantine' → 'levant')
df['guessed_dialect_clean'] = df['guessed_dialect_clean'].replace({
    'levantine': 'levant',
    'nile basin': 'egypt',  # In case this file has mixed regions
    'modern standard arabic': 'msa',
    # Add others if needed
})

df['dialect_clean'] = df['dialect'].str.strip().str.lower()

# ✅ Keep only rows where guessed matches declared dialect
matched_df = df[df['guessed_dialect_clean'] == df['dialect_clean']].copy()

# ✅ Rename column and finalize dialect label
matched_df = matched_df.rename(columns={'arabic_transcript': 'transcript'})
matched_df['dialect'] = 'Levant'  # Set final output dialect label

# ✅ Save to cleaned output
matched_df.to_csv("transcripts_youtube_videos_levant_module11_cleaned.csv", index=False, encoding='utf-8-sig')

print(f"✅ Saved {len(matched_df)} matching rows out of {len(df)}")
```
